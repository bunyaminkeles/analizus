```python
year1 =2023
year2 = 2023 # yıl2 yıl1 den büyük olmalı

for üni in üniversiteler[234:]:
    time.sleep(1)
    year1 =2023
    year2 = 2023 # yıl2 yıl1 den büyük olmalı    
    driver = get_driver(url)
    click_process(driver,üni)
    sonuç = sonuç_sayısı(driver)
    time.sleep(1)
    print(f"{üni} için sonuç sayısı: {sonuç}")
    
    if not sonuç:
        driver.close()
        continue
    if sonuç == 0:
        driver.close()
        continue
    elif sonuç < 2000:
        df = get_table(driver)
        en, tr = [], []
        for idx,rows in df.iterrows():
            e, t = get_abstract(driver,rows["tez_detay"],rows["tez_no"])
            en.append(e)
            tr.append(t)
        df['en'] = en
        df['tr'] = tr
        df.drop(columns=['tez_detay'], inplace=True)
        df.to_csv(f"data/{üni}.csv", index=False)
        driver.close()
        continue
    elif sonuç < 4000:
        driver.close()
        döngü = sonuç//2000+2
        for i in range(1,döngü):
            print(f"Toplam sonuç sayısı: 2000'den fazla olduğu için yıl aralığı seçiliyor")
            time.sleep(1)
            if i == döngü-1:
                yıl1 = 1900
            else:
                yıl1 = year1 - 12
            yıl2 = year2
            print(f"Yıl aralığı: {yıl1}-{yıl2}")
            driver = get_driver(url)
            click_process(driver,üni,yıl1=yıl1,yıl2=yıl2)
            sonuç = sonuç_sayısı(driver)
            if sonuç == 0:
                driver.close()
                continue
            print(f"{üni} için sonuç sayısı: {sonuç}")
            df = get_table(driver)
            en, tr = [], []
            for idx,rows in df.iloc[:].iterrows():
                e, t = get_abstract(driver,rows["tez_detay"],rows["tez_no"])
                en.append(e)
                tr.append(t)
            df['en'] = en
            df['tr'] = tr
            df.drop(columns=['tez_detay'], inplace=True)
            df.to_csv(f"data/{üni}_{yıl1}_{year2}.csv", index=False)
            driver.close()
            
            year1 = yıl1-1
            year2 = yıl1-1
        print(f"{üni} için işlem tamamlandı")
        print("-----------------------------------------------------------")
            
    else:
        driver.close()
        döngü = sonuç//1000+2
        for i in range(1,döngü):
            print(f"Toplam sonuç sayısı: 2000'den fazla olduğu için yıl aralığı seçiliyor")
            time.sleep(1)
            if i == döngü-1:
                yıl1 = 1900
            else:
                yıl1 = year1 - 6
            yıl2 = year2
            print(f"Yıl aralığı: {yıl1}-{yıl2}")
            driver = get_driver(url)
            click_process(driver,üni,yıl1=yıl1,yıl2=yıl2)
            sonuç = sonuç_sayısı(driver)
            if sonuç == 0:
                driver.close()
                continue
            else:
                print(f"{üni} için sonuç sayısı: {sonuç}")
                df = get_table(driver)
                en, tr = [], []
                for idx,rows in df.iloc[:].iterrows():
                    e, t = get_abstract(driver,rows["tez_detay"],rows["tez_no"])
                    en.append(e)
                    tr.append(t)
                df['en'] = en
                df['tr'] = tr
                df.drop(columns=['tez_detay'], inplace=True)
                df.to_csv(f"data/{üni}_{yıl1}_{year2}.csv", index=False)
                driver.close()
                
                year1 = yıl1-1
                year2 = yıl1-1
        print(f"{üni} için işlem tamamlandı")
        print("-----------------------------------------------------------")
            
            
            
```
